# Lecture 0.1: Intro to Pandas 🐼 DataFrames & Series for Machine Learning 

Before we dive into our first algorithm—**k-Nearest Neighbors (k-NN)**—we must establish a solid foundation in **Pandas**. 

In Machine Learning, real-world data rarely comes as raw arrays. Whether you are dealing with structured tabular datasets, labeled feature sets, or mixed-type raw files (CSVs, Excels), data is organized using **Pandas DataStructures (`DataFrame` & `Series`)**.

### Why Pandas for Machine Learning?
1. **Labeled Data & Feature Management**: Unlike NumPy's plain numerical arrays, Pandas allows us to store named columns (features) and index-aligned rows (samples), making feature selection and target isolation clear and readable.
2. **Seamless Data Handling & Preprocessing**: It provides powerful built-in functions to easily clean missing values, handle categorical features, rename variables, and merge datasets before model training.
3. **NumPy Integration**: Pandas is built directly on top of NumPy. You can effortlessly extract raw matrices using `.to_numpy()` or `.values` whenever your algorithm requires direct matrix operations (like calculating k-NN distance matrices).

---
### Comparison summary: NumPy vs Pandas

| Concept | NumPy (`ndarray`) | Pandas (`DataFrame` / `Series`) |
| :--- | :--- | :--- |
| **Primary Focus** | N-Dimensional numerical calculations & matrix math | Tabular data manipulation, cleaning & feature engineering |
| **Data Types** | Homogeneous (all elements must be the same type) | Heterogeneous (columns can hold integers, floats, strings, etc.) |
| **Indexing** | Integer-position based (`0, 1, 2...`) | Named labels (Column names) + Index labels / Positions (`loc`, `iloc`) |
| **Role in ML** | Distance metrics, loss functions, linear algebra | Loading raw datasets, Exploratory Data Analysis (EDA), missing data handling, encoding |

# What is EDA? (Exploratory Data Analysis) 🔍

 It is a critical foundational step in any **Machine Learning** or **Data Science** project, performed right after collecting data and before training any models.

### What is the Main Purpose of EDA?
1. **Understand the Data**: Inspect the structure, data types (`dtypes`), and dimensions (`shape`) of the dataset.
2. **Detect Data Issues**: Identify missing values (`Missing Values`), duplicate rows, or extreme anomalies (`Outliers`).
3. **Discover Patterns & Relationships**: Uncover correlations between variables (e.g., finding which features strongly influence the target).
4. **Define Features and Targets**: Clearly separate input feature columns (`Features - X`) from the target prediction column (`Target - Y`).

### The Role of Pandas in EDA:
**Pandas** is the core tool used to perform EDA efficiently using built-in methods such as:
- `df.info()`: To check data types and non-null value counts.
- `df.describe()`: To generate descriptive statistics (mean, standard deviation, min, max, percentiles).
- `df.isnull().sum()`: To calculate the total missing values per column.
- Often combined with visualization libraries like `Matplotlib` and `Seaborn` to visually plot and analyze these insights.


Note=> How `df.describe()` Handles Missing Values (`NaN`)
- **Automatic Exclusion**: `df.describe()` automatically ignores and excludes all `NaN` (missing) values when calculating statistics (like mean, std, min, max, and percentiles) for each column.
- **No Data Mutation**: It does **not** modify or drop rows from your original DataFrame; it simply computes the metrics based only on the valid (non-null) data points available in each column.
- **Impact on Count**: You can verify this by looking at the `count` row in the summary output. If a column has missing values, its `count` will be lower than the total number of rows in the DataFrame.

## 3. Data Selection, Slicing & Indexing

Selecting relevant features (columns) and samples (rows) is crucial for preparing your dataset for machine learning.

- `df['column_name']`: Select a single column as a Series.
- `df[['col1', 'col2']]`: Select multiple columns as a DataFrame
- `df.loc[row_label, col_label]`: Label-based selection.
- `df.iloc[row_index, col_index]`: Integer index-based selection.
# 🔍 `.loc` vs `.iloc`
| Feature / Aspect | `.loc` (Label-Based Indexing) | `.iloc` (Integer-Position-Based Indexing) |
| :--- | :--- | :--- |
| **Core Concept** | Accesses data using explicit **labels** of rows and columns (e.g., string names, dates, or custom index keys). | Accesses data strictly using integer **ordinal positions** from `0` to `n-1`, ignoring custom index names. |
| **Index Types Supported** | Strings, integers (if explicitly set as labels), DatetimeIndex, MultiIndex, and boolean arrays. | Integer positions only (`int`, lists of integers, integer slices), or boolean NumPy arrays. |
| **Slicing Endpoint** | **Inclusive** (`start:end` includes both `start` and `end` labels if they exist). | **Exclusive** (`start:end` includes `start` up to `end - 1`, following standard Python convention). |
| **Missing / Invalid Indexing** | Raises a `KeyError` if a specified row or column label does not exist in the DataFrame. | Raises an `IndexError` if the integer position exceeds the bounds of the axis dimensions. |
| **Boolean Filtering Support** | **Native & Direct**: Can directly accept boolean Series or lists aligned with the DataFrame index. | **Requires Conversion**: Direct boolean Series pass fails; requires converting to a NumPy array via `.to_numpy()` or `.values`. |
| **Callable / Lambda Support** | Fully supports callable functions (e.g., `df.loc[lambda df: df['Age'] > 25]`) for chaining. | Fully supports callable functions (e.g., `df.iloc[lambda df: [0, 2]]`) for pipeline processing. |
| **Performance (Speed)** | Slightly slower when resolving string labels, hash map lookups, or multi-indexes. | **Faster**: Operates directly on contiguous integer memory arrays and underlying NumPy matrix coordinates. |
| **Primary Machine Learning Use Case** | Extracting data subsets using feature names, class labels, or specific date-time ranges during preprocessing. | Slicing raw matrix shapes, splitting feature matrices (`X`) and targets (`y`) by numerical column positions. |

````python
import pandas as pd

# ==========================================
# LOADING REAL HOUSE PRICES DATASET FROM LIVE URL
# ==========================================
# Direct public URL for a real House Prices CSV dataset
url = "https://raw.githubusercontent.com/reisanar/datasets/master/kaggle-housing.csv"
# Read the dataset directly into a Pandas DataFrame
df = pd.read_csv(url)
# 2. EXPLORATORY DATA ANALYSIS (EDA)
print(f"\nDataset Shape (Rows, Columns): {df.shape}\n")
print("-------- 2. Dataset Info --------")
df.info()
print("\n------- 3. Statistical Summary --------")
print(df.describe())
````

````python
print("--- 1. First 5 Rows of Kaggle Housing Dataset ---")
df.head(5)
````

````python
# ==========================================
# 3. SELECTING FEATURES AND TARGET FOR ML
# ==========================================
# In Machine Learning tasks, we isolate our feature matrix (X) from the target price (y)

# Feature Matrix (Independent Variables) - renamed to 'house_features'
house_features = df[['area', 'bedrooms', 'bathrooms', 'stories']] # feature matrix (X)
print("--- House Features Matrix Head ---")
print(house_features.head(5))

# Target Vector (Dependent Variable / What we want to predict) 'target_price'
target_price = df['price'] # target vector (y)
print("\n--- Target Price Vector Head ---")
print(target_price.head(5))

# Checking the types of the feature matrix and target vector
print("\n--- Types of features and target ---")
print(type(house_features))  # Output: <class 'pandas.core.frame.DataFrame'>
print(type(target_price))   # Output: <class 'pandas.core.series.Series'>
````

````python
# ==================================================
# PART 1: ADVANCED EXAMPLES USING .loc (Label-Based)
# ==================================================

# Example 1: Select specific non-consecutive rows and specific columns by name
# Useful when you want to inspect particular samples (e.g., houses with index 0, 2, and 4)
print("--- .loc Example 1: Specific Rows and Columns ---")
print(df.loc[[0, 2, 4], ['price', 'area', 'bedrooms']]) 
# note [[0, 2, 4]] is a list of row labels, and ['price', 'area', 'bedrooms'] is a list of column labels
print("\n" + "_"*50 + "\n")


# Example 2: Inclusive Slicing (Range of rows)
# Notice that row index 2 IS included in the output (Unlike Python/iloc slicing)
print("--- .loc Example 2: Inclusive Range Slicing ---")
print(df.loc[0:2, ['area', 'price']])
print("\n" + "_"*50 + "\n")


# Select specific columns (price, bedrooms) only for houses where bedrooms > 3
print("--- .loc Example 3: Filtering with Conditions ---")
print(df.loc[df['bedrooms'] > 3, ['price', 'bedrooms', 'area']].head(3))
print("\n" + "_"*50 + "\n")

print("--- Range using .loc (Inclusive of end) ---")
# Example 4: Select rows from index 0 to index 3 (Row 3 WILL appear in the output)
# And select a range of columns from 'area' up to 'stories'
print(df.loc[0:3, 'area':'stories'])
print("\n" + "_"*50 + "\n")

# ======================================================
# PART 2: ADVANCED EXAMPLES USING .iloc (Position-Based)
# ======================================================

# Example 1: Select a single exact value using integer coordinates
# Row position 1, Column position 0 (first column)
print("--- .iloc Example 1: Single Cell Value ---")
print(df.iloc[1, 0])
print("\n" + "_"*50 + "\n")

# Example 2: Exclusive Slicing (Range of positions)
# Rows from 0 up to 3 (Excludes index 3, so rows 0, 1, 2) and columns 0 to 2
print("--- .iloc Example 2: Exclusive Slicing ---")
print(df.iloc[0:3, 0:3])
print("\n" + "_"*50 + "\n")

# Example 3: Selecting specific scattered rows and columns using integer lists
# Grabbing row positions 0, 5, 10 and column positions 0, 1, 3
print("--- .iloc Example 3: Custom Lists of Positions ---")
print(df.iloc[[0, 5, 10], [0, 1, 3]])


# Example 4: Range / Slicing WITH .iloc (Position-Based)
# Rule: The ending position IS EXCLUSIVE.

print("--- Range using .iloc (Exclusive of end) ---")
# Select row positions from 0 up to 3 (Rows 0, 1, and 2 will appear, but row 3 is EXCLUDED)
# And select column positions from 0 up to 3 (Columns 0, 1, and 2)
print(df.iloc[0:3, 0:3])
````

## 4. Conditional Filtering

Filtering rows based on specific conditions helps clean data and isolate subsets for evaluation.

````python
# ==========================================
# 4. CONDITIONAL FILTERING IN PANDAS
# ==========================================

# Condition 1: Filter houses where the number of bedrooms is strictly greater than 4
# ==============================================================================
# 1. Define the specific columns we want to display
selected_columns = ['area', 'bedrooms', 'bathrooms', 'stories', 'basement', 'price']

# 2. Filter for houses with 4 or more bedrooms, select specific columns, and group by 'bedrooms'
high_bedroom_houses = df.loc[df['bedrooms'] > 4, selected_columns]
print("--- Houses with More Than 4 Bedrooms ---")
print(high_bedroom_houses)  # Display only the feature columns for these houses
print(f"Number of houses with more than 4 bedrooms: {high_bedroom_houses['bedrooms'].count()}")
print("\n" + "-"*50 + "\n")

# Condition 2: Multiple Conditions using AND (&) and OR (|) operator
# ==============================================================================
# Note: Each condition must be wrapped in parentheses  () in Pandas.
# Let's filter houses that have bedrooms > 3 AND price > 10,000,000
luxury_large_houses = df[(df['bedrooms'] > 3) & (df['price'] > 10000000)]

print("--- 2. Multiple Conditions Filtering (AND operator) ---")
print(luxury_large_houses[selected_columns])  # Display only the feature columns for these houses
print(f"Total matching luxury houses: {luxury_large_houses.shape[0]}")
print("\n" + "-"*50 + "\n")
````

## 5. Handling Missing Data

Most Machine Learning algorithms cannot handle missing values (`NaN`). We must detect and either remove or impute them.

- `df.isnull().sum()`: Count missing values per column.
- `df.dropna()`: Remove rows with missing data.
- `df.fillna(value)`: Impute missing values with a constant or mean/median.

### Practical Example: Simulating and Fixing Missing Data

Since our original Kaggle housing dataset is clean and contains no missing values, we will **intentionally inject some `NaN` values** into a copy of the data for educational purposes to demonstrate how to detect and handle missing values in real-world machine learning projects.

````python
# ==============================================================================
# PRACTICAL EXAMPLE: Random Sampling with a Fixed Seed and Injecting NaNs
# ==============================================================================
import numpy as np

# 1. Take a random sample of 50 rows with a fixed seed for reproducibility,
# and reset the index so row positions are clean and sequential (0 to 49)
df_simulated = df.sample(n=50, random_state=42).reset_index(drop=True).copy()

# 2. Intentionally inject 3 NaN values into different rows and columns
df_simulated.loc[5, 'area'] = np.nan       # Inject NaN into row 5, 'area' column
df_simulated.loc[18, 'price'] = np.nan     # Inject NaN into row 18, 'price' column
df_simulated.loc[32, 'bedrooms'] = np.nan  # Inject NaN into row 32, 'bedrooms' column

print("--- Step 1: Sample Dataset (50 rows) with 3 Injected NaN Values ---")
# Display rows where we injected NaNs to verify
print(df_simulated[['area', 'bedrooms', 'price']].head(20))  # Display first 20 rows to include the injected NaNs
print("\n" + "-"*50 + "\n")
````

````python
# Create a copy for imputation
df_clean = df_simulated.copy()

# ==============================================================================
# DETECTING AND FIXING THE MISSING VALUES
# ==============================================================================

# Detect missing values by counting nulls per column
print("--- Step 1: Counting Missing Values ---")
print(df_simulated.isnull().sum())
print("\n" + "-"*50 + "\n")

# Impute (Fill) the missing values:
# - Fill missing 'price' with the column mean
df_clean['price'] = df_clean['price'].fillna(df_clean['price'].mean())

# - Fill missing 'area' with the column mean
df_clean['area'] = df_clean['area'].fillna(df_clean['area'].mean())

# - Fill missing 'bedrooms' with the column median
df_clean['bedrooms'] = df_clean['bedrooms'].fillna(df_clean['bedrooms'].median())


# Dropping rows with missing values (Alternative approach)
df_dropped = df_clean.dropna()

print("--- Step 2: Verification After Imputation (Total Missing Values) ---")
print(df_clean.isnull().sum())

print("\n" + "-"*50 + "\n")

print("--- Displaying Specific Values After Imputation ---")
print(f"Value in row 5, 'area' column: {df_clean.loc[5, 'area']}")
print(f"Value in row 18, 'price' column: {df_clean.loc[18, 'price']}")
print(f"Value in row 32, 'bedrooms' column: {df_clean.loc[32, 'bedrooms']}")
````

## 6. Data Cleaning & Feature Transformation

Preparing data for Machine Learning involves transforming columns, adding new features, and encoding categorical variables.

- `df.rename()`: Rename column headers.
- `df.drop()`: Drop unnecessary features.
- `df.astype()`: Convert data types.

````python
# ==========================================
# 6. SIMPLE DATA CLEANING & TRANSFORMATION
# ==========================================

# 1. Create a clean copy of the dataset to keep the original safe
df_cleaned = df.copy()
# 2. Rename a column (e.g., change 'stories' to 'num_floors')
df_cleaned = df_cleaned.rename(columns={'stories': 'num_floors'})
print("-> Column 'stories' renamed to 'num_floors'.")
# 3. Convert data type (e.g., convert 'area' to float)
df_cleaned['area'] = df_cleaned['area'].astype(float)
print("-> Column 'area' converted to float.")

# 4. Convert text columns (like 'mainroad') into numbers (0 and 1) using One-Hot Encoding

"""anthor approach is to use labelencoder like LabelEncoder() from sklearn.preprocessing, 
   but here we will use a simple map function for binary columns.""" 

if 'mainroad' in df_cleaned.columns:
    df_cleaned['mainroad'] = df_cleaned['mainroad'].map({'yes': 1, 'no': 0})
    print("-> 'mainroad' encoded quickly using map (yes=1, no=0)")

# Preview the first 3 rows of our clean dataset
print("\n--- Preview of Cleaned Dataset ---")
df_cleaned[['area', 'num_floors', 'bedrooms', 'bathrooms', 'mainroad', 'price']].head(3)
````

### 1. What is Data Mapping? (General Concept)

In data science, **Mapping** is the process of creating a direct correspondence or bridge between elements of one dataset category and a target format. 

Think of it like a translation dictionary: you take raw, qualitative values (such as text labels like `"Yes"`, `"No"`) and **map** them precisely to a quantitative scale (`1`, `0`) so that analytical systems and machine learning models can process them.

---

### 2. Why is Mapping Important?

- **Standardization:** Unifies inconsistent data formats into a clean, uniform structure.
- **Data Translation:** Converts words and categories into numbers required for statistical analysis.
- **Precision & Control:** Gives you 100% manual control over how each specific category translates.

---

### 3. Mapping in Python & Pandas (`.map()`)

In Python, we implement mapping using the `.map()` method on a Pandas Series. It allows us to transform values within the same column instantly.

#### **Function Parameters:**
The `Series.map()` function accepts the following main arguments:

1. **`arg` (Required):** The mapping correspondence. It can be:
   - A **Dictionary** (e.g., `{'yes': 1, 'no': 0}`) where keys are existing values and values are the new mapped targets.
   - A **Function** or a **Lambda expression** (e.g., `lambda x: x.lower()`) applied to each element.
   - A **Series** containing the corresponding mappings.
2. **`na_action` (Optional):** Controls how missing values (`NaN`) are handled.
   - Setting `na_action='ignore'` skips missing values and leaves them as `NaN` instead of trying to map them.

### 7. Grouping & Aggregation (`groupby`)

In exploratory data analysis and feature engineering, grouping data allows us to calculate summary statistics across categorical segments. This helps us uncover hidden patterns and relationships between features and the target variable.

- **`df.groupby('column')`**: Splits the data into groups based on unique categories.
- **`.mean()`, `.sum()`, `.count()`**: Standard aggregate functions applied to the groups.
- **`.agg()`**: Applies multiple aggregation functions simultaneously (e.g., mean, max, count).

````python
# ==========================================
# 7. GROUPING & AGGREGATION EXAMPLES
# ==========================================

# 1. Group by a categorical feature (e.g., 'num_floors') and find the average price
if 'price' in df_cleaned.columns and 'num_floors' in df_cleaned.columns:
    avg_price_by_floors = df_cleaned.groupby('num_floors')['price'].mean().reset_index()
    print("--- Average Price by Number of Floors ---")
    print(avg_price_by_floors)

# 2. Using .agg() to calculate multiple metrics (mean, count, max) for a group
if 'price' in df_cleaned.columns and 'mainroad' in df_cleaned.columns:
    summary_stats = df_cleaned.groupby('mainroad')['price'].agg(['mean', 'count', 'max'])
    print("\n--- Price Summary Statistics by Mainroad Access ---")
    print(summary_stats)
````

### 9. Integrating SQL with Pandas for Data Science

In real-world data science pipelines, data rarely lives in local CSV files. Instead, it resides in relational databases. Pandas integrates seamlessly with SQL, allowing us to query, filter, and aggregate large datasets directly into a Pandas `DataFrame` using `pd.read_sql()`.
 
- **SQL vs. Pandas Grouping:** SQL handles heavy data extraction, filtering (`WHERE`), and aggregation (`GROUP BY`) at the database level.
- **Pandas Role:** Takes the resulting clean dataset and prepares it for Machine Learning preprocessing (cleaning, encoding, splitting).

````python
# ==============================================================================
#9.SQL & PANDAS INTEGRATION (Practical Example)
# ==============================================================================
import sqlite3
import pandas as pd

# 1. Create a temporary in-memory SQLite database connection for demonstration
conn = sqlite3.connect(':memory:')

# Push our current df_cleaned to the SQL database as a table named 'housing_table'
df_cleaned.to_sql('housing_table', conn, index=False, if_exists='replace')

# 2. Write an SQL query mimicking GROUP BY and aggregate functions
sql_query = """
    SELECT num_floors, AVG(price) AS avg_price, COUNT(*) as total_houses
    FROM housing_table
    GROUP BY num_floors
    ORDER BY avg_price DESC;
"""

# 3. Read the SQL query results directly into a new Pandas DataFrame
df_from_sql = pd.read_sql(sql_query, conn)

# 4. Close the database connection
conn.close()

# Preview the resulting DataFrame from SQL
print("--- Aggregated Data Fetched via SQL into Pandas ---")
print(df_from_sql)
````
